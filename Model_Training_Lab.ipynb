```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
import joblib

print("🧠 Initializing AI Training Laboratory...")

# 1. GENERATE TRAINING DATA
# Since your database is new, we are simulating 5,000 days of historical stock data.
# We are teaching the AI to look at "Recent Return" and "Sentiment Score" to predict if the stock goes UP (1) or DOWN (0).
np.random.seed(42)
num_samples = 5000

# Simulated features (What the AI sees)
historical_returns = np.random.uniform(-0.05, 0.05, num_samples) # Daily momentum between -5% and +5%
historical_sentiment = np.random.uniform(-1, 1, num_samples)     # News sentiment between -1 (Bad) and 1 (Good)

# Simulated target (What actually happened next)
# Logic: If momentum is positive AND sentiment is positive, it's more likely to go up.
future_trend = ((historical_returns > 0) & (historical_sentiment > 0.2)).astype(int)

# Combine into a clean dataset
X = pd.DataFrame({"recent_return": historical_returns, "sentiment_score": historical_sentiment})
y = future_trend

# 2. TRAIN THE ALGORITHM
print("📊 Training Random Forest Classifier on 5,000 historical data points...")
ai_model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42)
ai_model.fit(X, y)

accuracy = ai_model.score(X, y)
print(f"🎯 Model Training Complete! Internal Accuracy: {accuracy * 100:.2f}%")

# 3. EXPORT THE BRAIN
filename = "stock_prediction.pkl"
joblib.dump(ai_model, filename)
print(f"💾 AI Brain successfully saved to your D: drive as '{filename}'!")
```
